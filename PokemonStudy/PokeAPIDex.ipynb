Starting with Pokemon

```python
import requests as r
import json
import csv
import pandas as pd
from pandas import json_normalize

#url to post
action_postURL = PokeDexURL
#get data from that URL
res = r.get(action_postURL)

#put the data coming in from res into values as json
values = res.json()
```

```python
#Define variables
PokeDex = [] 
MaxDexNumFound = 0
MaxDexNum = 0
CurrentNum = 0
PokeDexID = ''
PokemonName = ''

#Define headers
PokeDex.append(['PokeDexID|PokemonName'])

#Load table
while MaxDexNumFound == 0:
    #Check to see if PokeDexID exists 
    try:
        #If PokeDexID exists, add to extract
        PokeDexID =str(values['pokemon_entries'][CurrentNum]['entry_number'])
        PokemonName = values['pokemon_entries'][CurrentNum]['pokemon_species']['name'].title()
        PokeDex.append([PokeDexID + '|' + PokemonName])
        CurrentNum += 1
        #If PokeDexID does not exist, print the max PokeDexID found
    except:
        print('Max PokeDexID Found:' + str(CurrentNum))
        MaxDexNumFound = 1
        MaxDexNum = CurrentNum
```

```python
#Sample
PokeDex[0:10]
```

Pokemon Species

```python
CurrentSpeciesNum = 1
PreferredLanguage = 'en'
PokemonSpecies = []

#Define headers
PokemonSpecies.append(['PokeDexID|PokemonDescription'])

for iteration in range(CurrentSpeciesNum,MaxDexNum + 1):
    currentSpeciesURL = 'https://pokeapi.co/api/v2/pokemon-species/' + str(iteration)
    resSpecies = r.get(currentSpeciesURL)
    values = resSpecies.json()
    for GameAndLanguage in range(0,len(values['flavor_text_entries'])):
        if values['flavor_text_entries'][GameAndLanguage]['language']['name'] == PreferredLanguage:
            PokemonSpecies.append([str(iteration) + '|' + values['flavor_text_entries'][GameAndLanguage]['flavor_text'].replace('\n', ' ').replace('\x0c',' ')])
            iteration += 1
            break
```
