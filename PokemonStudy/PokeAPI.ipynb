```python
import requests as r
import json
from pandas import json_normalize

#url to post
action_postURL = 'https://pokeapi.co/api/v2/pokemon/132'

#get data from that URL
res = r.get(action_postURL)

#put the data coming in from res into values as json
values = res.json()
```

```python
values.keys()
```

```python
print('Pokemon #' + str(values['id']) + ': ' + values['name'])
```

```python
values['abilities']
```

```python
print('List of abilities:')
for i in range(0,len(values['abilities'])):
    print('- ' + values['abilities'][i]['ability']['name'])
```

```python
MaxDexNumFound = 0
CurrentDexNum = 1

while MaxDexNumFound == 0 :
    action_postURL = 'https://pokeapi.co/api/v2/pokemon/' + str(CurrentDexNum)
    res = r.get(action_postURL)
    if res.status_code == 404:
        MaxDexNumFound = 1
    else:
        values = res.json()
        print('Pokemon #' + str(values['id']) + ': ' + values['name'])
        CurrentDexNum += 1
```
